## Early-Stopping

Early-Stopping allows the model to stop before reaching the given amount of epochs, when the monitored loss does not improve anymore.\
Here is an example.

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = ""
import sys
sys.path.append("../../")

import matplotlib.pyplot as plt
import tensorflow as tf
tf.config.run_functions_eagerly(True)
```

```python
from choice_learn.datasets import load_swissmetro
from choice_learn.models import MultiNomialLogit

dataset = load_swissmetro()

callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="train_loss", patience=20, min_delta=0, mode="min")
]
model = MultiNomialLogit(epochs=1_000, optimizer="adam", callbacks=callbacks, lr=1e-3, batch_size=-1)
hist = model.fit(dataset)

plt.plot(hist["train_loss"])
plt.xlabel("Epochs")
plt.ylabel("Train Loss")
plt.show()
```
